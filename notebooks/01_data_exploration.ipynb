```python
apartments = df[df["tip"] == "apartament"]
```

```python
apartments["Zona"].value_counts()
```

```python
apartments[
    (apartments["an_constructie"] > 2022) |
    (apartments["an_constructie"] < 1900)
][["Zona", "Denumire strada", "an_constructie"]]
```

```python
apartments[
    (apartments["S utila"] <= 10) |
    (apartments["S utila"] > 300)
][
    ["Zona", "Denumire strada", "S utila", "Pret tranzactie LEI", "an_constructie"]
].sort_values("S utila")
```

```python
apartments["S utila"].quantile(
    [0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
)
```

```python
apartments = apartments.copy()

apartments["pret_per_sqm"] = (
    apartments["Pret tranzactie LEI"] /
    apartments["S utila"]
)
```

```python
apartments["pret_per_sqm"].describe()
```

```python
apartments[
    [
        "Zona",
        "Denumire strada",
        "S utila",
        "Pret tranzactie LEI",
        "pret_per_sqm",
        "an_constructie"
    ]
].sort_values(
    "pret_per_sqm",
    ascending=False
).head(10)
```
